```python
from typing import List
```

```python
nums1 = [1,2,3,0,0,0]
nums2 = [2,5,6]
```

```python
class solution:
    def merge(self,nums1:List[int],m:int,nums2:List[int],n:int) -> None:

        #last index
        last=m+n-1

        #reverse 
        while m > 0 and n > 0:
            if nums1[m-1] > nums2[n-1]:
                nums1[last]=nums1[m-1]
                m-=1
            else:
                nums1[last]=nums2[n-1]
                n-=1

            last-=1

        #leftover

        while n > 0:
            nums1[last]=nums2[n-1]
            n,last=n-1,last-1
```
